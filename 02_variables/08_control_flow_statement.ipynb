## If, elif, eles

```python
age = 6
if age >= 18:
    print("You are eligible to vote.")
elif age <= 5:
    print("You are still a toddler.")
elif age >=6 and age <= 10:
    print("You shoud enrol in a school with Form-B.")
else:
    print("Abhi chotay ho ap.")
```

## Nested If statments

```python
age = 18
license = True
if age >= 18:
    if license:
        print("You are eligible to drive.")
    else:
        print("You are not eligible to drive, You need a license.")
else:
    print("You are not eligible to drive.")
```

## While Loops

```python
cup_of_chai = 0
while cup_of_chai < 5:
    print("I want more chai.")
    cup_of_chai += 1

print("I have had enough chai.")
```

```python
cup_of_chai = 0
while cup_of_chai < 10:
    print("I want more chai.")
    cup_of_chai += 1
print("I want had enough chai")
```

## Break and continue

```python
cups_of_chai = 0

while cups_of_chai < 5:
    cups_of_chai += 1
    print("I want more chai.")
    if cups_of_chai == 3:
        print("Too much chai, stopping now.")
        break
    print("I have had", cups_of_chai, "cups of chai.")
```

## Continue

```python
cups_of_chai = 0

while cups_of_chai < 5:
    cups_of_chai += 1
    print("I want more chai.")
    if cups_of_chai == 2:
        print("Too much chai, stopping now.")
        continue
    print("I have had", cups_of_chai, "cups of chai.")
```

## Range function in loops

## infinite loops

```python
while True:
    print("This loop will run forever")
    break # you need to be careful while running this loop
    
```

```python
count = 0 

while True:
    count += 1
    print("This loop will run forever")
    if count == 10:
        break
```

## Pass statment

```python
cups_of_chai = 0

while cups_of_chai < 5:
    cups_of_chai += 1
    print("I want more chai.")
    if cups_of_chai == 3:
        print("Too much chai, stopping now.")
        pass
    print("I have had", cups_of_chai, "cups of chai.")
```
