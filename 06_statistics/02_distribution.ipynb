```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="whitegrid")

# Generate data
np.random.seed(42)
normal_data = np.random.normal(loc=0, scale=1, size=1000)        # Mean=0, Std=1
uniform_data = np.random.uniform(low=0, high=10, size=1000)       # Range 0 to 10
exponential_data = np.random.exponential(scale=1, size=1000)      # Lambda=1

# Create plots
plt.figure(figsize=(15, 4))

# Normal Distribution
plt.subplot(1, 3, 1)
sns.histplot(normal_data, kde=True, bins=30, color='skyblue')
plt.title('Normal Distribution')

# Uniform Distribution
plt.subplot(1, 3, 2)
sns.histplot(uniform_data, kde=True, bins=30, color='orange')
plt.title('Uniform Distribution')

# Exponential Distribution
plt.subplot(1, 3, 3)
sns.histplot(exponential_data, kde=True, bins=30, color='green')
plt.title('Exponential Distribution')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn style
sns.set(style="whitegrid")

# Generate data
np.random.seed(42)
normal_data = np.random.normal(loc=0, scale=1, size=1000)        # mean=0, std=1
uniform_data = np.random.uniform(low=0, high=10, size=1000)       # range: 0 to 10
exponential_data = np.random.exponential(scale=1, size=1000)      # scale = 1/lambda

# Create subplots
plt.figure(figsize=(15, 4))

# Normal Distribution
plt.subplot(1, 3, 1)
sns.histplot(normal_data, bins=30, color='skyblue')
plt.title('Normal Distribution')

# Uniform Distribution
plt.subplot(1, 3, 2)
sns.histplot(uniform_data, bins=30, color='orange')
plt.title('Uniform Distribution')

# Exponential Distribution
plt.subplot(1, 3, 3)
sns.histplot(exponential_data, bins=30, color='green')
plt.title('Exponential Distribution')

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.stats import skew, kurtosis

# Generation example data
np.random.seed
data = np.random.normal(0, 1, 1000) # Normal distribution 

# Calculate skewness and kurtosis
data_skewness = skew(data)
data_kurtosis = kurtosis(data, fisher=True) # 'fisher=True' returns excess kurtosis

# Displaying the results
print(f"Skewness ogf the dataset: {data_skewness}")
print(f"Kurtosis of the dataset: {data_kurtosis}")
```

```python
sns.histplot(data, kde=True)
```

```python
# how to test the normal distribution of the data 

# H0: The data is normally distributed
# H1: The data is not normally distributed

# shpiro-wilk test
# D'Agsotino's K^2 test
# Anderson-Darling test
```

```python
# shapiro-wilk test to test normality of the data
stats.shapiro(data)
print(f"Shapiro-Wilk test statistic: {stats.shapiro(data)[0]}")
print(f"Shapiro-Wilk test p-value: {stats.shapiro(data)[1]}")

```
