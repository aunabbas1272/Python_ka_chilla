# Anova
ANOVA (Analysis of Variance) is a statistical method developed by Sir Ronald A. Fisher in the early 1920s to test whether there are significant differences between the means of three or more independent groups. Fisher introduced this technique while working at the Rothamsted Experimental Station in England to analyze agricultural data, and he first published it in his landmark book, Statistical Methods for Research Workers in 1925. ANOVA works by comparing the variance between groups to the variance within groups using the F-statistic, and it's widely used in fields such as medicine, psychology, business, and scientific research. Fisher, often called the "Father of Modern Statistics," also contributed to experimental design, hypothesis testing, and many foundational ideas in modern statistics.

## 1. One-Way ANOVA

Purpose: Tests the effect of one independent variable (with 2 or more groups) on a single dependent variable.
Example: Comparing average test scores of students taught using 3 different teaching methods.


## 2. Two-Way ANOVA
Purpose: Tests the effect of two independent variables and their interaction on one dependent variable.
Example: Comparing the effect of teaching method and gender on student performance.


## 3. N-Way ANOVA 
N-Way ANOVA (also called Factorial ANOVA) is a statistical method used to examine the effects of two or more independent variables (factors) on a single dependent variable, including the interactions between the factors.
