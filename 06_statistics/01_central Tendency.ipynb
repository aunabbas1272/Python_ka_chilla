```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Genrating a dataset with potential outliers
np.random.seed(0)
data = np.random.normal(100, 20, 200)
data = np.append(data, [150, 200]) #Adding potential outliers

# Caculating IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

#Defining the outliers criteria (1.5 times the IQR)
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Identifying the outliers
outliers = (data < lower_bound) | (data > upper_bound)

# Removing the outliers
data_no_outliers = data[~outliers]

# Plotting the original data with outliers
plt.figure(figsize=(14,6))
plt.subplot(1,2,1)
sns.boxplot(data=data)
plt.title('Original Data with Outliers')

# Plotting the data after removing outliers
plt.subplot(1,2,2)
sns.boxplot(data=data_no_outliers)
plt.title('Data after Removing Outliers')

plt.show()
```
